```python
import pandas as pd
```

```python
df = pd.read_csv('StudentPerformanceFactors.csv')

df.info()
```

```python
# describe() # Generates descriptive statistics of the data, including count, mean, std, min, max, and quantiles.
df.describe()
```

```python
# Include only object data  type
df.describe(include=['object'])
```

```python
# Include only number data type 
df.describe(include=['number'])
```

```python
# Include all data ttype 
df.describe(include='all')
```

```python
# selection specific percentiles
df.describe(percentiles=[0.1, 0.5, 0.9])
```

```python
#  Using describe on a non-numerical column 
df['Parental_Education_Level'].describe()
```

```python
# Descriptive Statistics

df['Exam_Score'].mean() # Computes the mean of the data. The column should be a numeric ddatatype
df['Exam_Score'].median() # Computes the median of the data. The column should be a numeric ddatatype
df.mode() # Returns the mode(s) of the data.
df['Exam_Score'].std() # Computes the standard deviation of the data.
df['Exam_Score'].var() # Computes the variance of the data.
df['Exam_Score'].min() # Returns the minimum value of the data.
df['Exam_Score'].max() # Returns the maximum value of the data.
df['Exam_Score'].sum() # Returns the sum of the values.
df['Exam_Score'].prod() # Returns the product of the values.
df['Exam_Score'].quantile() # Computes the value at the given quantile. The default is 0.5 or 50%
df.count() # Counts the number of non#null values.
df['Exam_Score'].cumsum() # Returns the cumulative sum.
df['Exam_Score'].cumprod() # Returns the cumulative product.
df['Exam_Score'].cummax() # Returns the cumulative maximum.
df['Exam_Score'].cummin() # Returns the cumulative minimum.


# Correlation and Covariance

df['Exam_Score'].corr(df['Attendance'])  # Computes the pairwise correlation of columns, excluding NA/null values.
df['Exam_Score'].cov(df['Attendance']) # Computes the pairwise covariance of columns, excluding NA/null values.



# Other Statistical Methods

df['Exam_Score'].skew() # Computes the skewness of the data (a measure of asymmetry).
df['Exam_Score'].kurt() # Computes the kurtosis (a measure of tailedness) of the data.
df['Exam_Score'].sem() # Computes the standard error of the mean.
df['Exam_Score'].abs() # Computes the absolute value of each element.
```
