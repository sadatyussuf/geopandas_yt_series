```python
import pandas as pd 
import numpy as np
```

```python
rows, cols = 10,5

data = np.random.randn(rows,cols)


df= pd.DataFrame(data,columns=[f'col_{i}' for i in range(cols)])

# Randamly assign Nan values
nan_vals = np.random.choice(df.size,size=int(df.size*0.2),replace=False)

df.values.ravel()[nan_vals] = np.nan

# nan_vals
df
```
