```python
import pandas as pd 
import numpy as np
```

# Replacing Missing Data

## Using replacing on a DataFrame 

```python
# Sample data
data = {
    'Column1': ['10', 'NA', 'N/A', '-', None, ' ', '25'],
    'Column2': ['100', '200', 'n/a', 'None', 'null', '-', '300'],
    'Column3': ['50', 'n/a', '-', None, 'N/A', 'null', '75'],
    'Column4': ['-', 'None', 'null', '50', '100', 'n/a', None],
    'Column5': ['25', '50', '100', 'n/a', '-', 'None', '150']
}


df = pd.DataFrame(data)

# Replacing missing value indicators with NaN
df.replace(['NA', 'N/A', 'n/a', 'None', 'null', '-', ' ',None], np.nan, inplace=True)

df
```

##  Using na_values on Read_CSV

```python
df = pd.read_csv('sample_data.csv', na_values=['NA', 'N/A', 'n/a', 'None', 'null', '-', ' ',None])
print(df.shape)
df
```

# How Missing Data Behaves in different DataTypes

## Missing Data in Numeric Series or DataFrame(column)

```python
# Assigning a None or np.nan to numeric Series or Dataframe(column) defaults to NaN

df_num = pd.Series([11.0, np.nan, 5, 9, 7,2.8 ,None])

df_num
```

## Missing Data in DateTime Series or DataFrame(column)

```python
# Assigning a None or np.nan to datetime Series or Dataframe(column) defaults to NaT

df_dt = pd.Series(['2024-01-01', None, '2024-12-26', np.nan,'2025-01-03'])

df_dt = pd.to_datetime(df_dt)

df_dt
```

## Missing Data in Object Series or DataFrame(column)

```python
# None and np.nan are treated as generic missing values in an Object Series or DataFrame(column).

df_obj = pd.Series([57,'sadat', None, 'z',pd.to_datetime('2025-01-03'), np.nan,20.89])

df_obj
```

# Filling Missing Data

## Filling with a Scalar Value using fillna

```python
df.fillna(0)
# df.fillna('NA')
```

## Filling with a Series using fillna

```python
# print(df.mean())
# print(type(df.mean()))
# print('---------------------------')
df.fillna(df.mean())
```

## Filling cells using Forward Fill (ffill)

```python
df.ffill()
# df.fillna(method='ffill') is another way of achieving the same results
```

## Filling cells using Backward Fill (bfill)

```python
df.bfill()
```
