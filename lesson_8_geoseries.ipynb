```python
import geopandas as gpd 
from shapely.geometry import Point, Polygon , LineString
```

```python
# Create a GeoSeries with Shapely geometries

one_point = Point(-2.7828, 9.73056)

one_linestring = LineString([
    (-2.7828, 9.73056),   
    (-2.9812, 9.2319)
])


one_polygon = Polygon([
    (-2.7828, 9.73056),  
    (-2.9812, 9.73056),
    (-2.9812, 9.2319),
    (-2.7828, 9.2319)  
])


geoseries_point = gpd.GeoSeries(data=one_point, name='geometry', crs='EPSG:4326')


geoseries_linestring = gpd.GeoSeries(data=one_linestring, name='geometry', crs='EPSG:4326')


geoseries_polygon = gpd.GeoSeries(data=one_polygon, name="geometry", crs="EPSG:4326")
```

```python
geoseries_point, geoseries_linestring, geoseries_polygon
```

```python
geoseries_point.plot()
geoseries_linestring.plot()
geoseries_polygon.plot()
```

```python
#  Create a GeoSeries from list of Shapely Geometries

points = [
    Point(-2.7828, 9.73056),
    Point(-2.9812, 9.2319)
]

linestrings = [
    LineString([
        (-2.7828, 9.73056),   
        (-2.9812, 9.2319)
    ]),
    LineString([
        (-2.7, 9.1),   
        (-3.2812, 9.8319)
    ])
]


polygons = [
    Polygon([
            (-2.7828, 9.73056),   
            (-2.9812, 9.2319),
            (-2.7, 9.1),   
            (-3.2812, 9.8319)
    ]),

    Polygon([
            (-2.6928, 9.6606),   
            (-2.9, 9.206),
            (-2.7, 9.1),   
            (-3.0182, 9.6319)
    ])
]


geoseries_points = gpd.GeoSeries(points, crs='EPSG:4326')
geoseries_lines = gpd.GeoSeries(linestrings, crs='EPSG:4326')
geoseries_polygons = gpd.GeoSeries(polygons, crs='EPSG:4326')
```

```python
geoseries_points, geoseries_lines, geoseries_polygons
```

```python
geoseries_points.plot()
geoseries_lines.plot()
geoseries_polygons.plot(alpha=0.5)
```

```python
#  Create a GeoSeries from list of different Shapely Geometries

one_point = Point(-2.7828, 9.73056)

one_linestring = LineString([
    (-2.7, 9.1),   
    (-3.2812, 9.8319)  
])

one_polygon = Polygon([
    (-2.6928, 9.6606),   
    (-3.0182, 9.6319),
    (-2.9, 9.206),
    (-2.7, 9.1),   
    (-2.6928, 9.6606),   
])

geoseries_collection = gpd.GeoSeries(
    [one_point, one_linestring, one_polygon], crs='EPSG:4326'
)

geoseries_collection
```

```python
import matplotlib.pyplot as plt

# geoseries_collection.plot()

# create a figure and axis
fig, ax = plt.subplots(figsize=(6, 6))

# give each geometry its own style
styles = [
    {"color": "red", "markersize": 80},                 # point
    {"color": "blue", "linewidth": 3},                  # line
    {"facecolor": "none", "edgecolor": "green", "linewidth": 2}   # polygon (no fill)
]

# plot each geometry separately
for geom, style in zip(geoseries_collection, styles):
    gpd.GeoSeries([geom]).plot(ax=ax, **style)

plt.show()
```

```python
# Create a GeoSeries from WKT strings

wkt_points = [
"POINT (-2.7828 9.73056)",
"POINT (-1.55971 8.80719)",
]

wkts_lines = [
"LINESTRING (-2.78367 9.73142, -2.78148 9.73291, -2.77812 9.73428)",
"LINESTRING (-2.78154 9.72987, -2.78021 9.73169, -2.77778 9.73612)",
]

wkts_polygons = [
"POLYGON ((-2.78 9.73, -2.77 9.73, -2.77 9.74, -2.78 9.74, -2.78 9.73))",
"POLYGON ((-2.78092 9.73131, -2.76869 9.73156, -2.76908 9.74082, -2.78111 9.74038, -2.78092 9.73131))",
]


geoseries_points_wkt = gpd.GeoSeries.from_wkt(wkt_points, crs="EPSG:4326")
geoseries_lines_wkt = gpd.GeoSeries.from_wkt(wkts_lines, crs="EPSG:4326")
geoseries_polygons_wkt = gpd.GeoSeries.from_wkt(wkts_polygons, crs="EPSG:4326")
```

```python
geoseries_points_wkt.plot()
geoseries_lines_wkt.plot()
geoseries_polygons_wkt.plot(alpha=0.5)
```
